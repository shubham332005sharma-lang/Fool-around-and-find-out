```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
```

```python
x = np.array([2, 7, 13, 18, 24, 29, 34, 40, 45, 49])
y = np.array([8, 10, 17, 15, 25, 24, 31, 36, 39, 46])


m=.8
c=5
y_line = m*x + c

plot=plt.scatter(x,y)

# draw the prediction line
plt.axline((0, c), slope=m, color='red', linestyle='--', label=f'y = {m}x + {c}')

# vertical lines from predicted to actual points 
plt.vlines(x,y,y_line,color="black" , linestyle = ":",label="Residual",zorder=0)

plt.xlim(left=0)
plt.ylim(bottom=0)

plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.legend()
plt.show()
```
