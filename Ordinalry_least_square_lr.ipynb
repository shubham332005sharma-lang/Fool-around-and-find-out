```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
```

```python
x = np.array([2, 7, 13, 18, 24, 29, 34, 40, 45, 49])
y = np.array([8, 10, 17, 15, 25, 24, 31, 36, 39, 46])


m=.8
c=5
y_line = m*x + c

plot=plt.scatter(x,y)

# draw the prediction line
plt.axline((0, c), slope=m, color='red', linestyle='--', label=f'y = {m}x + {c}')

# vertical lines from predicted to actual points 
plt.vlines(x,y,y_line,color="black" , linestyle = ":",label="Residual",zorder=0)

plt.xlim(left=0)
plt.ylim(bottom=0)

plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.legend()
plt.show()
```

```python
residual = y - (m * x + c)
print(residual)
#pros of using numpy arrays
rss = sum(residual**2)
print("Residual Error is : ",rss)
```

```python
slope = sum((x-(sum(x)/10))*(y-(sum(y)/10)))/sum((x-(sum(x)/10))**2)
print(slope)
intercept= (sum(y)/10)- slope*(sum(x)/10)
print(intercept)
# In ODLS we differentiate the cost function once with respect to slope and then with intercept which at minimum =0 gives us the points . 
# The thing to note here is that the calculation of the slope is complicated and has a sizing factor problem , so if we scale this to million rows.. it fails . 
# The Gradient Descent Method comes in clutch in such cases , it uses low computational power , converges quickly ( subjective to learning rate ) , which we will see tomorrow . 
```
